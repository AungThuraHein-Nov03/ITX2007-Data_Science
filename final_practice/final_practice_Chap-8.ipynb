```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import uniform
import statsmodels.formula.api as smf
```

```python
ad_conversion = pd.read_csv("../Week9/ad_conversion.csv")
print(ad_conversion.head())
```

```python
sns.regplot(x = "n_impressions", y = "n_clicks", data = ad_conversion)
```

```python
ad_conversion["qdrt_n_impressions"] = ad_conversion["n_impressions"] ** 0.25
ad_conversion["qdrt_n_clicks"] = ad_conversion["n_clicks"] ** 0.25

sns.regplot(x = "qdrt_n_impressions", y = "qdrt_n_clicks", data = ad_conversion)
```

```python
mdl_click_vs_impressions = smf.ols("qdrt_n_clicks ~ qdrt_n_impressions", data = ad_conversion).fit()
print(mdl_click_vs_impressions.params)
```

```python
explanatory_data = pd.DataFrame({"n_impressions": np.arange(0, 3000001, 500000)})
explanatory_data["qdrt_n_impressions"] = explanatory_data["n_impressions"] ** 0.25
prediction_data = explanatory_data.copy()
prediction_data["qdrt_n_clicks"] = mdl_click_vs_impressions.predict(explanatory_data)

print(prediction_data)
```

```python
prediction_data["n_clicks"] = prediction_data["qdrt_n_clicks"] ** 4
sns.regplot(x = "n_impressions", y = "n_clicks", data=ad_conversion)
sns.scatterplot(x = "n_impressions", y = "n_clicks", data = prediction_data, color="red", label="Predicted")
plt.show()
```

```python
mse = mdl_click_vs_impressions.mse_resid
print(mse)
rse = np.sqrt(mse)
print(rse)
```
